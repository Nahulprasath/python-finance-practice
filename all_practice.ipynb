```python
# pattern 1 — increasing
n = int(input("Enter the number: "))
r = 0
while r != n + 1:
    print(r * "*")
    r = r + 1

# pattern 2 — decreasing while loop
n = int(input("Enter the number: "))
while n > 0:
    print(n * "*")
    n = n - 1

# pattern 3 — for loop increasing
for i in range(1, 6):
    print(i * "*")

# pattern 4 — for loop decreasing
r = int(input("Enter the num: "))
for i in range(r, 0, -1):
    print(i * "*")

# pattern 5 — while loop decreasing
i = int(input("Enter: "))
while i != 0:
    print(i * "*")
    i -= 1
```

```python
def daily_return(prev, curr):
    return round((curr - prev) / prev * 100, 2)

names = ["Reliance", "TCS", "Infosys", "HDFC"]
prev  = [2350, 3150, 1520, 1700]
curr  = [2400, 3200, 1480, 1650]

for i in range(len(names)):
    print(names[i], ":", daily_return(prev[i], curr[i]), "%")
```

```python
import pandas as pd

data = {
    "Stock":  ["Reliance","TCS","Infosys","HDFC","Wipro","ONGC","ITC","Bajaj","Axis","HCL"],
    "Price":  [2400, 3200, 1500, 1650, 420, 890, 340, 4200, 980, 1100],
    "Change": [1.2, -0.5, 2.1, -1.3, 0.8, -2.4, 1.5, -0.9, 1.8, -1.1],
    "Sector": ["Energy","IT","IT","Finance","IT","Energy","FMCG","Finance","Finance","IT"],
    "Risk":   ["Low","High","Medium","Low","High","High","Low","Medium","Medium","Low"]
}
df = pd.DataFrame(data)
df["Value"] = df["Price"] * 100

print(df.groupby("Sector").agg({
    "Value":  ["sum", "mean", "count"],
    "Change": ["mean"]
}))

def gain(change):
    if change > 0:
        return "Gaining"
    else:
        return "Losing"

df["Status"] = df["Change"].apply(gain)
print(df)
```

```python
import numpy as np

tcs_prices = np.array([3150, 3200, 3180, 3250, 3220, 3280])

returns = np.diff(tcs_prices) / tcs_prices[:-1] * 100

print("Daily returns %:", np.round(returns, 2))
print("Avg daily return:", round(np.mean(returns), 2), "%")
print("Volatility (std dev):", round(np.std(returns), 2), "%")

if np.mean(returns) > 0:
    print("Good week")
else:
    print("Bad week")
```

```python
import numpy as np

names  = ["Reliance", "TCS", "Infosys", "HDFC", "Wipro"]
prices = np.array([2400, 3200, 1500, 1650, 420])
prev   = np.array([2350, 3150, 1520, 1700, 415])

# step 1 — calculate returns
returns = (prices - prev) / prev * 100

# step 2 — signals
signals = np.where(returns > 2,  "STRONG BUY",
          np.where(returns < -2, "HIGH RISK", "WATCH"))

# step 3 — print each stock
print("=" * 50)
print("DAILY RISK REPORT")
print("=" * 50)
for i in range(len(names)):
    print(names[i], "→ ₹", prices[i],
          "→ Return:", round(returns[i], 2), "%",
          "→", signals[i])

# step 4 — portfolio summary
print("=" * 50)
print("Portfolio avg return:", round(np.mean(returns), 2), "%")
print("Portfolio risk (std dev):", round(np.std(returns), 2), "%")

# step 5 — count high risk
count = 0
for s in signals:
    if s == "HIGH RISK":
        count += 1
print("HIGH RISK stocks:", count)
print("=" * 50)
```
